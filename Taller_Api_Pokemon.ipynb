```python
import requests
import pandas as pd
```

```python
# Consulta a la API
url = "https://pokeapi.co/api/v2/pokemon?limit=10"
response = requests.get(url)

# Verificar respuesta
if response.status_code == 200:
    data = response.json()
    pokemons = data['results']
else:
    print("Error en la consulta:", response.status_code)

# Convertir a DataFrame
df = pd.DataFrame(pokemons)
df.head()
```

```python
# Obtener detalles adicionales
details = []
for p in pokemons:
    r = requests.get(p['url'])
    if r.status_code == 200:
        info = r.json()
        details.append({
            'name': info['name'],
            'height': info['height'],
            'weight': info['weight'],
            'types': ', '.join([t['type']['name'] for t in info['types']])
        })

# Crear DataFrame final
df_final = pd.DataFrame(details)
df_final.head()
```

```python
df_final.to_csv('pokemon_dataset.csv', index=False)
print("Archivo CSV generado correctamente.")
```
