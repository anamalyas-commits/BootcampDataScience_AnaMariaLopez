```python
# Apartir de la exloración inicial evidencimos que las columnas "Age","Cabin" y "Embarked", contienen valores nulos. Por esta razón, antes del análisis es necesario realizar limpieza de datps 
```

```python
# 2. Limpieza de datos: Se realiza limpieza del dataset para tratar valores faltantes y tener la base lista para el análisis
```

```python
# Eliminar filas duplicadas en caso que existan

df = df.drop_duplicates()
```

```python
# Llenar edades vacías con el promedio.

df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
# Rellenar valores faltantes de Embarked con el valor mas frecuente.

df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
# Eliminar la columna "Cabin" por que tiene demasiados nullos

df = df.drop(columns=["Cabin"])
```

```python
df.isnull().sum()
```

```python
# 3. Transformación de datos: Se crean nuevas variables que facilitaran el analisis de los datos.
```

```python
# Se crea la columna "GrupoEdad", para clasificar a los pasajeros entre menores de edad y adultos.

df["GrupoEdad"] = df["Age"].apply(lambda x: "Menor de edad" if x < 18 else "Adulto")
print(df[["Age", "GrupoEdad"]].head())
```

```python
# 4. Preguntas.
# Pregunta 1: ¿La tasa de supervivencia fue diferente entre hombre y mujeres?

## Calcular la tasa promedio de supervivenvia por sexo.

df.groupby("Sex")["Survived"].mean()
```

```python
# Cuenta cuantos hombre y mujeres hay.

df.groupby("Sex")["PassengerId"].count()
```

```python
# Uso de filtros.

mujeres_sobrevivieron = df[(df["Sex"] == "female") & (df["Survived"] == 1)].shape[0]
hombres_sobrevivieron = df[(df["Sex"] == "male") & (df["Survived"] == 1)].shape[0]

print("Mujeres que sobrevivieron", mujeres_sobrevivieron)
print("Hombres que sobrevivieron", hombres_sobrevivieron)
```

```python
# Se puede evidenciar que la tasa de supervivencia fue mayor en mujeres que en hombres. Esto indica que el sexo estuvo fuerte mente relacionado con la probabilidad de sobrevivir.
```

```python
# Pregunta 2: ¿La clase del pasajero influyó en la supervivencia y en la tarifa pagada?

# Calcular la supervivencia promedio por clase.

df.groupby("Pclass")["Survived"].mean()
```

```python
# Calcular la tarifa promedio por clase.

df.groupby("Pclass")["Fare"].mean()
```

```python
df.groupby("Pclass")[["Survived","Fare"]].mean()
```

```python
# Uso de flitros

primera_clase = df[df["Pclass"] == 1].shape[0]
segunda_clase = df[df["Pclass"] == 2].shape[0]
tercera_clase = df[df["Pclass"] == 3].shape[0]

print("Pasajeros en primera clase:", primera_clase)
print("Pasajeros en segunda clase:", segunda_clase)
print("Pasajeros en tercera clase:", tercera_clase)
```

```python
# Los resultados muestran que los pasajeros en primera clase tuvieron una mayo tasa de supervivencia y pagaron tarifas mas altas. Esto indica que la clase social y económica del pasajero influyó en sus posibilidades de sobrevivir.  
```

```python
# Pregunta 3: ¿Los menores de edad tuvieron una supervivencia diferente a la de los adultos? 

df.groupby("GrupoEdad")["Survived"].mean()
```

```python
df["GrupoEdad"].value_counts()
```

```python
menores = df[df["GrupoEdad"] == "Menor de edad"].shape[0]
adultos = df[df["GrupoEdad"] == "Adulto"].shape[0]

print("Menores de edad", menores)
print("Adultos", adultos)
```

```python
# Con esos resultados sí se puede afirmar que la edad influyó en la probabilidad de supervivencia. ya que los menores de edad tuvieron una tasa de de sobrevivencia 53% y los adultos una tada de 36%  
```
